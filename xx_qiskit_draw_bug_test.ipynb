```python
import os
import numpy as np
from qiskit import ClassicalRegister, QuantumRegister, QuantumCircuit
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
### Sample circuit
def qmodel(qubit_no, before_layers, after_layers, add_meas=False):   
    qr = QuantumRegister(qubit_no, 'q')
    ansatz = QuantumCircuit(qr, name='ansatz')
    for q in range(qubit_no):
        ansatz.rx(q*0.1, q)
    ansatz.measure_all()
    return ansatz
```

```python
### Plot X, y
def xy_plot(X, y):
    plt.title(f'Sample Xy plot')
    plt.xlabel(f'X')
    plt.ylabel('y')
    plt.plot(X, y, color='black')
    plt.show()
```

```python
def fun(x):
    return (np.sin(5.0 * x) + 0.5*np.sin(8.0 * x)) / 4 + 0.5

lb = 0; ub = 1; samples=100
X = np.linspace(lb, ub, num=samples)
y = fun(X)
```

## Show the bug

```python
### Print the plot
xy_plot(X, y)
```

```python
### Create a sample model and draw it
#   Problem happens only with 'mpl'
model = qmodel(5, 2, 1, add_meas=True)
model.draw(output='mpl') 
```

```python
### Print the plot immediately after the circuit draw
xy_plot(X, y)
```

```python
### Print the plot again and again
xy_plot(X, y)
```
